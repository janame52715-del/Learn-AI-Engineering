```python
# 销售数据分析

## 实验目的

本实验用于学习：

1. 使用 Pandas 创建和读取 CSV 文件。
2. 使用 DataFrame 展示数据。
3. 使用 Matplotlib 绘制数据图表。
```

```python
import pandas as pd

data = pd.DataFrame({
    "month": ["January", "February", "March", "April", "May"],
    "sales": [120, 150, 180, 160, 220]
})

data.to_csv(
    "./sample_sales.csv",
    index=False
)
```

```python
df = pd.read_csv("./sample_sales.csv")

display(df)
```

```python
import matplotlib.pyplot as plt

df.plot(
    x="month",
    y="sales",
    kind="bar",
    figsize=(8, 4)
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```
